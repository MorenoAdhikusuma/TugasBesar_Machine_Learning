```python
df['harga'] = (
    df['harga']
    .astype(str)
    .str.replace(r'[^\d.]', '', regex=True)  # keeps digits and decimal point only
    .replace('', '0')
    .astype(float)
    .astype(int)
)
print(df.dtypes)
```

```python
import numpy as np
```

```python
conditions = [
    df['harga'] < 3_000_000,
    (df['harga'] >= 3_000_000) & (df['harga'] < 8_000_000),
    (df['harga'] >= 8_000_000) & (df['harga'] < 15_000_000),
    (df['harga'] >= 15_000_000) & (df['harga'] <= 30_000_000),
    df['harga'] > 30_000_000,
]

labels = [
    'Sangat Murah',
    'Murah',
    'Menengah',
    'Mahal',
    'Sangat Mahal',
]

df['kategori_harga'] = np.select(conditions, labels, default='Tidak Diketahui')
```

```python
print(df['kategori_harga'].value_counts(dropna=False))

# Spot-check values aligned with each condition tier
for low, high in [
    (0, 2_999_999),            # Sangat Murah (< 3,000,000)
    (3_000_000, 7_999_999),    # Murah (3,000,000 - 7,999,999)
    (8_000_000, 14_999_999),   # Menengah (8,000,000 - 14,999,999)
    (15_000_000, 30_000_000),  # Mahal (15,000,000 - 30,000,000)
    (30_000_001, 75_000_000),  # Sangat Mahal (> 30,000,000)
]:
    subset = df[df['harga'].between(low, high)][['nama', 'harga', 'kategori_harga']]
    if subset.empty:
        print(f"No rows between {low:,} and {high:,}")
    else:
        display(subset.head())
```
